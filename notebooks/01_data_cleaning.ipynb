# 5. Cleaning

```python
def clean_dataset(df):
    """
    This function takes in a dataframe and performs the following cleaning steps:
    - Fills any missing values with 0
    - Removes any rows with missing values
    - Removes any duplicate rows
    - Standardizes column names by converting them to lowercase and replacing spaces with underscores
    - Converts all string values to lowercase
    - Resets the index of the dataframe
    """
    df = df.fillna(0)
    df = df.dropna()
    df = df.drop_duplicates()
    df.columns = df.columns.str.lower().str.replace(' ', '_').str.strip()
    df = df.apply(lambda x: x.str.strip() if x.dtype == "object" else x)
    df = df.reset_index(drop=True)
    return df   






print(clean_dataset(df))


```
